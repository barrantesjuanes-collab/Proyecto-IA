```python
df['FECHA_ACCIDENTE'] = pd.to_datetime(df['FECHA_ACCIDENTE'], format='%m/%Y', errors='coerce')

df['año'] = df['FECHA_ACCIDENTE'].dt.year
df['mes'] = df['FECHA_ACCIDENTE'].dt.month
```

```python
df['fatal'] = df['GRAVEDAD_ACCIDENTE'].apply(lambda x: 1 if x == 'CON MUERTOS' else 0)
```

```python
df['EDAD_VEHICULO'] = pd.to_numeric(df['EDAD_VEHICULO'], errors='coerce')
df = df.dropna(subset=['EDAD_VEHICULO'])
df = df[df['EDAD_VEHICULO'] < 50]
```

```python
# Ensure 'fatal' column exists
if 'fatal' not in df.columns and 'GRAVEDAD_ACCIDENTE' in df.columns:
    df['fatal'] = df['GRAVEDAD_ACCIDENTE'].apply(lambda x: 1 if x == 'CON MUERTOS' else 0)

sns.countplot(x='fatal', data=df)
plt.title("Distribución de accidentes fatales vs no fatales")
plt.xlabel("Fatal (0 = no, 1 = sí)")
plt.ylabel("Número de accidentes")
plt.show()
```
